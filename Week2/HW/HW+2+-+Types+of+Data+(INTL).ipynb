## Task 2

Given a dictionary of students with their scores, find the average score of each student across all subjects and identify the student with the highest average score.

```python
students_grades = {
    "John": [85, 90, 78],
    "Alice": [88, 92, 80],
    "Bob": [75, 85, 72],
    "Diana": [90, 95, 94],
    "Charlie": [70, 65, 80]
}
```

```python
# Calculate and print the average grade for each student
avg_grades = {}
for person in students_grades:
    avg = sum(students_grades[person]) / len(students_grades[person])
    print(f"{person} : {avg:.2f}")
    avg_grades[person] = avg
```

```python
# Find the student with the highest average grade
highavg = None
for person in avg_grades :
    if highavg is None or avg_grades[highavg] < avg_grades[person] :
        highavg = person
print(f"{highavg} is the highest average grade and the average grade is {avg_grades[highavg]}") 
```

# Part 3 : Working with Matplotlib

Plot a graph showing the average salary of employees in each department to compare the average salaries across departments.

```python
# Load CSV and create DataFrame
data = pd.read_csv('employee_data.csv')
df = pd.DataFrame(data)
```

```python
# Calculate average MonthlyRate by Department
avg_MonthlyRate = df.groupby("Department").agg({
    "MonthlyRate" : "mean"
})
print(avg_MonthlyRate)
```

```python
plt.figure(figsize=(10, 6))
plt.bar(avg_MonthlyRate.index, avg_MonthlyRate['MonthlyRate'])
plt.xlabel('Department')
plt.ylabel('Mean Monthly Rate')
plt.title('Mean Monthly Rate by Department')
plt.grid(axis='y', linestyle='--', alpha=0.7) 
```
