```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
```

1. Load all pokemon data (from previous Homework)

```python
# write your code here
df = pd.read_csv("pokemon.csv")
df
```

2. Are there any missing values? If so, in which column? \
Ans: Yes, there are. column which has missing values is 'Type 2'.

```python
df.isnull().sum()
```

3. Calculate the average 'Attack' of each pokemon 'Type 1'. Which type has the most Attack stats? \
Ans: Dragon type has the most Attack stats that is 112.125.

```python
# write your code here
print(df.groupby('Type 1')["Attack"].mean())
```

4. Aggregate count of each Pokemon Type 1

```python
# write your code here
typy1count = df.groupby('Type 1').size()
print(typy1count)
```

5. Create a visualization that show proportion of each pokemon Type 1 in the dataset\
**Hint:** plotly 

```python
# write your code here
fig = px.pie(
    df,
    names='Type 1',
    title='Proportion of Each Pokemon Type 1',
    hole=0.4
)

fig.update_traces(textposition='inside', textinfo='percent+label')

fig.show()
```

6. Create a line plot of 'Attack', 'Defense', and 'HP' of Bulbasaur evolution. \
Bulbasaur -> Ivysaur -> Venusaur

```python
# write your code here
evolution = df.iloc[:3]

plt.figure(figsize=(10, 6))
plt.plot(evolution['Name'], evolution['Attack'], marker='o', label='Attack')
plt.plot(evolution['Name'], evolution['Defense'], marker='s', label='Defense')
plt.plot(evolution['Name'], evolution['HP'], marker='^', label='HP')

plt.title('Bulbasaur evolution')
plt.xlabel('Pokemon')
plt.ylabel('Base Stat')
plt.legend()
plt.show()
```

7. Create a histogram of Pokemon total stats

```python
# write your code here
plt.figure(figsize=(10, 6))

plt.hist(df['Total'], edgecolor='black')

plt.title('Pokemon Total Stats')
plt.xlabel('Total Stat Value')
plt.ylabel('Number of Pokemon')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```
